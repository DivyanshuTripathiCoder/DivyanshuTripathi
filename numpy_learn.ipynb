**Ranges and Linespaces**

---


*   range ---->  np.arange(start, ending-1, step)
*   linespace-->np.linspace(starting_value, ending_value,number of values)






```python
#for making range wali array by using "np.arange"
range_wali_array=np.arange(0,15,2)
print(range_wali_array)
#for making line space wali array using "np.linspace"
linespace_wali_array=  np.linspace(0,1,4)
print(linespace_wali_array)
```

**SPECIAL ARRAYS:** we can build any dim of array which have zeros anad ones only

---


*   zeros --> np.zeros((rows,coloumn))
*   ones  --> np.ones((rows,coloumn))


*   identity(square matrix)--> np.eye(rows **/** column)




```python
#zeros wali array
zero_wali= np.zeros((3,3))
print(zero_wali)
print()
#one wali array
one_wali_array= np.ones((3,3))
print(one_wali_array)
print()
#identity
one_and_zero= np.eye(5)
print(one_and_zero)
```

```python
#RESHAPING AN ARRAY and FLATENING ARRAY
arr=np.arange(6)# making an array using arange function
print(arr) #printing the array
new=arr.reshape(2,3) #reshaping the array using reshape function
print(new) #printing the reshaped array
new_new=new.flatten() # flattening the reshaped array using flatten() function
print(new_new)#printing the flattned array
```

**MATHEMATICAL OPERATIONS:**



---







```python
vals=np.array([4,9,16]) #making an array named as vals
vals= vals+2 #adding 1 to all the vals's element(vectorised operation),any operation can be done(+,-,*,**,/,//,%..).
print(vals) #printing the vals array
#---> some basic fun of operations : np.sqrt, np.power, np.add
print(np.sqrt(vals))# printing the sqrt using np.sqrt() function
print(np.power([vals],[2]))# printing the power using np.power([array],[power]) function
print(np.add([vals],[vals+1]))# printing the addition of 2 array using np.add([array],[array2]) function
```

**INDEXING AND FILTERING AND SORTING**

---



```python
arr=np.array([1,5,7,2,8]) #creting an array
filtered=arr[arr>4]#putting the values bigger than 4 in a new var named filtered
print(filtered)# printing the filtered array
indices=np.where(arr>4) # putting the inxed of the elements greater than 4 into a var indices
print(indices)# printing all those indices
sorted=np.sort(arr)# sorting the array using np.sort(array) function
print("sorted array:",sorted)# printing the sorted array
arg_sort=np.argsort(arr)#sorting the indices and putting in a var named arg_sort using function -> np.argsort(array)
print("sorted indices:", arg_sort)#printing the sorted indices
```
